```python
from typing import TypeVar, Generic
T = TypeVar("T")

class Inventory(Generic[T]):
    def __init__(self):
        self.items = []

    def store(self,item: T):
        self.items.append(item)

    def retrieve(self):
        return self.items

    def display(self):
        for item in self.items:
            print(f"Type: {type(item).__name__},Value: {item}")

#product names
product_names = Inventory[str]()
product_names.store("Laptop")
product_names.store("Keyboard")

#Stock quantities
stock = Inventory[int]()
stock.store(50)
stock.store(120)

#Product prices
prices = Inventory[float]()
prices.store(3500.50)
prices.store(120.75)

#Serial number batches
serials = Inventory[list]()
serials.store(["SN001","SN002","SN003"])
serials.store(["SN004","SN005"])

print("Product Names")
product_names.display()

print("\nStock Quantities")
stock.display()

print("\nProduct Prices")
prices.display()

print("\nSerials Number Batches")
serials.display()

```
