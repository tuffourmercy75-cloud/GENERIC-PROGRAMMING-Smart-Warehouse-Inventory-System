```python
from typing import Generic, TypeVar
T = TypeVar("T")
#Generic class
class Storage(Generic[T]):
    def __init__(self):
        self.item =None

    def store(self,item:T):
        self.item = item

    def retrieve(self) -> T:
        return self.item

#Integer
int_storage = Storage[int]()
int_storage.store(100)
print("Integer:", int_storage.retrieve())

#String
str_storage = Storage[str]()
str_storage.store("Mercy")
print("String:", str_storage.retrieve())

#list
list_storage = Storage[list]()
list_storage.store([10,20,30,40])
print("List:", list_storage.retrieve())


```
